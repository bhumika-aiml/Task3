```python

data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha", "Aman", "Priya", "Rahul", None],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "IT", "Finance"],
    "Age": [21, 22, 20, 25, 21, None, 20, 24],
    "Salary": [35000, 30000, 40000, 45000, 35000, 32000, -5000, 45000],
    "City": ["Delhi", "Mumbai", "Delhi", "Bhopal", "Delhi", "Mumbai", "Delhi", "Bhopal"]
}

df = pd.DataFrame(data)

df.to_csv("employees.csv", index=False)

print("CSV file created successfully!")
```

```python
df = pd.read_csv("employees.csv")

display(df)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["Name"])
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df.isnull().sum())
```

```python
print(df)
```

```python
it_employees = df[df["Department"] == "IT"]

print(it_employees)
```

```python
high_salary = df[df["Salary"] > 35000]

print(high_salary)
```

```python
delhi_employees = df[df["City"] == "Delhi"]

print(delhi_employees)
```

```python
department_group = df.groupby("Department")

print(department_group["Salary"].mean())
```

```python
summary = df.groupby("Department")["Salary"].agg(
    ["count", "mean", "min", "max"]
)

print(summary)
```

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print("Average Salary by Department:")
print(avg_salary)
```

```python
print("Average Age:", df["Age"].mean())
```

```python
print(df["Department"].value_counts())
```

```python
print("Cleaned Dataset:")
display(df)
```

```python
df.to_csv("cleaned_employees.csv", index=False)

print("Cleaned dataset saved successfully!")
```

The dataset was analyzed using Pandas. Missing values were identified and handled, while incorrect salary values were removed. The data was filtered based on department, salary, and city. Grouping and aggregation were used to calculate department-wise employee counts and average salaries. The analysis helps understand salary patterns and employee distribution across departments
